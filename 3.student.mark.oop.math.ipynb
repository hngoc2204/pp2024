```python
import math
import numpy as np

class Student:
    def __init__(self, id, name, dob, gpa):
        # __ for private, _ for protected
        self.__id= id
        self.__name= name
        self.__dob= dob
        self.__gpa= 0
    
    def get_id(self):
        return self.__id
    
    def get_name(self):
        return self.__name
    
    def get_dob(self):
        return self.__dob

    def get_gpa(self):
        return self.__gpa
        
    def set_gpa(self, gpa):
        self.__gpa= gpa
    
    def __lt__(self, other):  # For sorting by GPA
        return self.__gpa < other.__gpa



class Course:
    def __init__(self, id, name, ects):
        self.__id= id
        self.__name= name
        self.__ects= ects
    
    def get_id(self):
        return self.__id

    def get_name(self):
        return self.__name   

    def get_ects(self):
        return self.__ects


class StudentMark:
    def __init__(self, course_id, student_id, score):
        self.__course_id= course_id
        self.__student_id= student_id
        self.__score= score

    def get_course_id(self):
        return self.__course_id

    def get_student_id(self):
        return self.__student_id

    def get_score(self):
        return self.__score
        


def student_num():
    num= int(input("The number of students: "))
    return num

def course_num():
    num= int(input("The number of courses: "))
    return num


class School:

    def __init__(self, studentNum, courseNum):
        self.__students= []
        self.__courses= []
        self.__marks= []
        self.__gpa= []
        self.__studentNum= studentNum
        self.__courseNum= courseNum

    def input_student_info(self):
        for i in range ((int)(self.__studentNum)): 
            student_id = input("Student ID: ")
            name = input("Name of the student: ")
            dob = input("DoB of the student: ")
            student = Student(student_id, name, dob, 0)
            self.__students.append(student)
    
    def input_course_info(self):
        for i in range ((int)(self.__courseNum)):
            course_id = input("Course ID: ")
            name = input("Name of the course: ")
            ects = input("Number of ECTS of the course: ")
            course = Course(course_id, name, ects)
            self.__courses.append(course)
            
    def input_mark(self):
        for course in self.__courses:
            for student in self.__students:
                m= float(input("Mark of the student "+ str(student.get_id()) + " in this course "+ str(course.get_id()) + " : "))
                mark= StudentMark(course.get_id(), student.get_id(), m) #original là để math.floor(m)
                self.__marks.append(mark)
                
    def list_course(self):
        for course in self.__courses:
            print("Course ID: "+ str(course.get_id()))
            print("Course name: "+ str(course.get_name()))
            print("Number of ECTS: "+ str(course.get_ects()))
        
    def list_student(self):
        for student in self.__students:
            print("Student ID: "+ str(student.get_id()))
            print("Student name: "+ str(student.get_name()))
            print("DoB: "+ str(student.get_dob()))
            print("GPA: "+ str(student.get_gpa()))
    
    def list_mark(self):
        course_id= str(input("Course ID to show marks: "))
        for mark in self.__marks:
            if str(mark.get_course_id()) == course_id: 
                print("Student ID: "+ str(mark.get_student_id()))
                print("Mark: "+ str(mark.get_score()))
                
    def calculate_gpa(self, student_id):
        weighted_sum =0
        sum_credits= 0
        for mark in self.__marks:
            for course in self.__courses:
                if str(course.get_id()) == str(mark.get_course_id()):
                    if str(mark.get_student_id())== student_id:
                        weighted_sum += (int) (course.get_ects()) * (int) (mark.get_score())
                        sum_credits += (int) (course.get_ects())
        gpa= weighted_sum/sum_credits
        return round(gpa,2)
        
                        
    def assign_gpa_to_students(self):
        for student in self.__students:
            gpa= self.calculate_gpa(student.get_id())
            student.set_gpa(gpa)
            
    def sort_by_gpa(self):
        self.assign_gpa_to_students()
        self.__students.sort(reverse= True)


num1= student_num()
num2= course_num()
s= School(num1, num2)

s.input_student_info()
s.input_course_info()
s.input_mark()
s.sort_by_gpa()
s.list_student()
```
